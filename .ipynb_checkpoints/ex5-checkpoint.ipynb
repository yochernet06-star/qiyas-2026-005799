```python
import requests
from bs4 import BeautifulSoup

html_doc = requests.get('https://www.worldometers.info/coronavirus/')
soup = BeautifulSoup(html_doc.content, 'html.parser')

table_content = soup.find('table')

table_head = table_content.find('thead')
table_body = table_content.find('tbody')
table_row = table_body.find_all('tr')

file = open("yoniso.csv", "a", encoding="utf-8")

# Write headers
headers = table_head.find_all('th')
for head in headers:
    file.write(head.text.strip() + ",")
file.write("\n")

# Write rows
for row in table_row:
    cols = row.find_all('td')

    for col in cols:
        data = col.text.strip().replace(",", "")
         data = col.text.strip().replace("+", "")
        file.write(data + ",")

    file.write("\n")

file.close()

```
