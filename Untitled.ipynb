```python
import requests
from bs4 import BeautifulSoup
import csv
import time
from datetime import datetime
import re
from urllib.parse import urljoin, urlparse
import json

class BBCAmharicScraper:
    def __init__(self):
        self.base_url = "https://www.bbc.com/amharic"
        self.categories = [
            {
                'name': 'zena',
                'url': 'https://www.bbc.com/amharic',
                'display_name': 'ዜና'
            },
            {
                'name': 'ethiopia',
                'url': 'https://www.bbc.com/amharic/topics/c7zp57r92v5t',
                'display_name': 'ኢትዮጵያ'
            },
            {
                'name': 'politica',
                'url': 'https://www.bbc.com/amharic/topics/cg7265pj1jvt',
                'display_name': 'ፖለቲካ'
            },
            {
                'name': 'setoch',
                'url': 'https://www.bbc.com/amharic/topics/cnq681w4dq6t',
                'display_name': 'ሴቶች'
            },
            {
                'name': 'tena',
                'url': 'https://www.bbc.com/amharic/topics/cxnyk76p0q2t',
                'display_name': 'ጤና'
            },
            {
                'name': 'sport',
                'url': 'https://www.bbc.com/amharic/topics/cdr56g2x71dt',
                'display_name': 'ስፖርት'
            },
            {
                'name': 'tebeb',
                'url': 'https://www.bbc.com/amharic/topics/cr50yvqzzwpt',
                'display_name': 'ጥበብ'
            },
            {
                'name': 'technology',
                'url': 'https://www.bbc.com/amharic/topics/c06gq8wx467t',
                'display_name': 'ቴክኖሎጂ'
            },
            {
                'name': 'metane_haet',
                'url': 'https://www.bbc.com/amharic/topics/cnq6815jj3xt',
                'display_name': 'ምጣኔ ሃብት'
            }
        ]
        self.headers = {
            'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36',
            'Accept-Language': 'am,en-US;q=0.9,en;q=0.8',
        }
        self.all_articles = []
        self.article_links = set()  # To avoid duplicates

    def fetch_page(self, url):
        """Fetch page content with error handling"""
        try:
            response = requests.get(url, headers=self.headers, timeout=15)
            response.raise_for_status()
            return BeautifulSoup(response.content, 'html.parser')
        except requests.RequestException as e:
            print(f"Error fetching {url}: {e}")
            return None

    def extract_article_info(self, article_element, category_name, category_display):
        """Extract information from a single article element"""
        try:
            # Extract title - try multiple selectors
            title = ""
            title_selectors = ['h1', 'h2', 'h3', 'h4', '[class*="title"]', '[class*="heading"]']
            for selector in title_selectors:
                title_elem = article_element.find(selector)
                if title_elem:
                    title = title_elem.get_text(strip=True)
                    break
            
            if not title:
                # Try to get from link text
                link_elem = article_element.find('a')
                if link_elem:
                    title = link_elem.get_text(strip=True)
            
            if not title:
                return None
            
            # Extract link
            link = ""
            link_elem = article_element.find('a')
            if link_elem:
                link = link_elem.get('href', '')
                if link and not link.startswith('http'):
                    link = urljoin(self.base_url, link)
            
            # Skip if duplicate link
            if link in self.article_links:
                return None
            if link:
                self.article_links.add(link)
            
            # Extract date
            date = ""
            date_selectors = ['time', '[class*="date"]', '[class*="time"]', '[class*="published"]', '[data-timestamp]']
            for selector in date_selectors:
                date_elem = article_element.find(selector)
                if date_elem:
                    date = date_elem.get_text(strip=True)
                    if date:
                        break
            
            # Extract content summary
            content = ""
            content_selectors = ['p', '[class*="summary"]', '[class*="description"]', '[class*="excerpt"]', '[class*="content"]']
            for selector in content_selectors:
                content_elem = article_element.find(selector)
                if content_elem:
                    content = content_elem.get_text(strip=True)
                    if content:
                        break
            
            # Extract image
            img_url = ""
            img_elem = article_element.find('img')
            if img_elem:
                img_url = img_elem.get('src', '')
                if img_url and not img_url.startswith('http'):
                    img_url = urljoin(self.base_url, img_url)
            
            return {
                'category': category_name,
                'category_display': category_display,
                'title': title,
                'link': link,
                'date': date,
                'content': content[:500] if content else '',  # Limit content length
                'image_url': img_url
            }
        except Exception as e:
            print(f"Error extracting article info: {e}")
            return None

    def scrape_category(self, category):
        """Scrape a specific category page"""
        category_name = category['name']
        category_display = category['display_name']
        url = category['url']
        
        print(f"Scraping category: {category_display} ({category_name})")
        soup = self.fetch_page(url)
        
        if not soup:
            print(f"Failed to fetch {url}")
            return []
        
        articles = []
        
        # Find article elements - try different selectors based on BBC's structure
        article_selectors = [
            'article',
            'li[class*="article"]',
            'div[class*="story"]',
            'div[class*="card"]',
            'div[class*="teaser"]',
            'div[class*="promo"]',
            'div[data-entityid]',
            'li[class*="promo"]'
        ]
        
        article_elements = []
        for selector in article_selectors:
            elements = soup.select(selector)
            if elements:
                article_elements = elements
                break
        
        # If no articles found, look for any divs with potential article content
        if not article_elements:
            for div in soup.find_all('div', recursive=True):
                class_list = div.get('class', [])
                class_str = ' '.join(class_list) if class_list else ''
                if any(keyword in class_str.lower() for keyword in ['story', 'article', 'promo', 'card', 'teaser']):
                    article_elements.append(div)
        
        # If still no articles, look for list items
        if not article_elements:
            for li in soup.find_all('li', recursive=True):
                class_list = li.get('class', [])
                class_str = ' '.join(class_list) if class_list else ''
                if any(keyword in class_str.lower() for keyword in ['story', 'article', 'promo']):
                    article_elements.append(li)
        
        print(f"Found {len(article_elements)} potential article elements in {category_display}")
        
        for element in article_elements[:20]:  # Limit to 20 articles per category
            article_info = self.extract_article_info(element, category_name, category_display)
            if article_info:
                articles.append(article_info)
        
        print(f"Extracted {len(articles)} articles from {category_display}")
        return articles

    def scrape_all_categories(self):
        """Scrape all categories"""
        print(f"Starting BBC Amharic scraping at {datetime.now()}")
        print(f"Total categories: {len(self.categories)}")
        
        for category in self.categories:
            category_articles = self.scrape_category(category)
            self.all_articles.extend(category_articles)
            print(f"Total articles from {category['display_name']}: {len(category_articles)}")
            time.sleep(2)  # Be respectful to the server
        
        print(f"Total unique articles scraped: {len(self.all_articles)}")
        return self.all_articles

    def save_to_csv(self, filename=None):
        """Save scraped articles to CSV file"""
        if not filename:
            filename = f"bbc_amharic_news_{datetime.now().strftime('%Y%m%d_%H%M%S')}.csv"
        
        if not self.all_articles:
            print("No articles to save!")
            return
        
        with open(filename, 'w', newline='', encoding='utf-8-sig') as csvfile:
            fieldnames = ['category', 'category_display', 'title', 'link', 'date', 'content', 'image_url']
            writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
            
            writer.writeheader()
            for article in self.all_articles:
                writer.writerow(article)
        
        print(f"Data saved to {filename}")
        return filename

    def save_to_json(self, filename=None):
        """Save scraped articles to JSON file"""
        if not filename:
            filename = f"bbc_amharic_news_{datetime.now().strftime('%Y%m%d_%H%M%S')}.json"
        
        if not self.all_articles:
            print("No articles to save!")
            return
        
        with open(filename, 'w', encoding='utf-8') as jsonfile:
            json.dump(self.all_articles, jsonfile, ensure_ascii=False, indent=2)
        
        print(f"Data saved to {filename}")
        return filename

    def get_category_stats(self):
        """Get statistics about scraped articles per category"""
        stats = {}
        for article in self.all_articles:
            category = article['category']
            if category not in stats:
                stats[category] = 0
            stats[category] += 1
        return stats

    def print_summary(self):
        """Print a summary of the scraping results"""
        print("\n" + "="*60)
        print("SCRAPING SUMMARY")
        print("="*60)
        print(f"Total articles scraped: {len(self.all_articles)}")
        print("\nArticles per category:")
        stats = self.get_category_stats()
        for category, count in stats.items():
            # Find display name
            display_name = category
            for cat in self.categories:
                if cat['name'] == category:
                    display_name = cat['display_name']
                    break
            print(f"  {display_name}: {count}")
        print("="*60)

def main():
    scraper = BBCAmharicScraper()
    
    # Scrape all categories
    scraper.scrape_all_categories()
    
    # Print summary
    scraper.print_summary()
    
    # Save to CSV
    csv_file = scraper.save_to_csv()
    
    # Also save to JSON for easier parsing
    json_file = scraper.save_to_json()
    
    # Show sample of data
    if scraper.all_articles:
        print("\nSample article:")
        sample = scraper.all_articles[0]
        for key, value in sample.items():
            print(f"  {key}: {value[:100] if value and len(str(value)) > 100 else value}...")
    
    return scraper.all_articles

if __name__ == "__main__":
    main()
```

```python
import requests
from bs4 import BeautifulSoup
import csv
import time
from datetime import datetime
from urllib.parse import urljoin
import re

class BBCAmharicFullScraper:
    def __init__(self):
        self.base_url = "https://www.bbc.com/amharic"
        self.categories = [
            {
                'name': 'zena',
                'url': 'https://www.bbc.com/amharic',
                'display_name': 'ዜና'
            },
            {
                'name': 'ethiopia',
                'url': 'https://www.bbc.com/amharic/topics/c7zp57r92v5t',
                'display_name': 'ኢትዮጵያ'
            },
            {
                'name': 'politica',
                'url': 'https://www.bbc.com/amharic/topics/cg7265pj1jvt',
                'display_name': 'ፖለቲካ'
            },
            {
                'name': 'setoch',
                'url': 'https://www.bbc.com/amharic/topics/cnq681w4dq6t',
                'display_name': 'ሴቶች'
            },
            {
                'name': 'tena',
                'url': 'https://www.bbc.com/amharic/topics/cxnyk76p0q2t',
                'display_name': 'ጤና'
            },
            {
                'name': 'sport',
                'url': 'https://www.bbc.com/amharic/topics/cdr56g2x71dt',
                'display_name': 'ስፖርት'
            },
            {
                'name': 'tebeb',
                'url': 'https://www.bbc.com/amharic/topics/cr50yvqzzwpt',
                'display_name': 'ጥበብ'
            },
            {
                'name': 'technology',
                'url': 'https://www.bbc.com/amharic/topics/c06gq8wx467t',
                'display_name': 'ቴክኖሎጂ'
            },
            {
                'name': 'metane_haet',
                'url': 'https://www.bbc.com/amharic/topics/cnq6815jj3xt',
                'display_name': 'ምጣኔ ሃብት'
            }
        ]
        self.headers = {
            'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36',
            'Accept-Language': 'am,en-US;q=0.9,en;q=0.8',
        }
        self.all_articles = []
        self.article_links = set()
        self.total_articles = 0

    def fetch_page(self, url):
        """Fetch page content with error handling and retry"""
        try:
            response = requests.get(url, headers=self.headers, timeout=15)
            response.raise_for_status()
            return BeautifulSoup(response.content, 'html.parser')
        except requests.RequestException as e:
            print(f"  Error fetching {url}: {e}")
            return None

    def clean_text(self, text):
        """Clean text by removing extra whitespace and newlines"""
        if not text:
            return ""
        return ' '.join(text.split())

    def extract_full_article_content(self, article_url):
        """Visit the article page and extract the full content"""
        print(f"    Fetching full article: {article_url[:60]}...")
        
        soup = self.fetch_page(article_url)
        if not soup:
            return "", ""
        
        # Extract full content - try multiple selectors
        content = ""
        content_selectors = [
            'div[class*="story-body"]',
            'div[class*="article-body"]',
            'div[class*="content"]',
            'div[class*="story"]',
            'article',
            'main',
            '[class*="bbc"]'
        ]
        
        for selector in content_selectors:
            content_elem = soup.select(selector)
            if content_elem:
                # Get all paragraphs within the content
                paragraphs = content_elem[0].find_all('p')
                if paragraphs:
                    content = ' '.join([self.clean_text(p.get_text(strip=True)) for p in paragraphs])
                    if content:
                        break
        
        # If no paragraphs found, try getting all text from the article
        if not content:
            article_main = soup.find('article') or soup.find('main')
            if article_main:
                paragraphs = article_main.find_all('p')
                if paragraphs:
                    content = ' '.join([self.clean_text(p.get_text(strip=True)) for p in paragraphs])
        
        # Extract image from the article
        img_url = ""
        img_elem = soup.find('img')
        if img_elem:
            img_url = img_elem.get('src', '')
            if img_url and not img_url.startswith('http'):
                img_url = urljoin(self.base_url, img_url)
        
        return content, img_url

    def extract_articles_from_page(self, soup, category_name, category_display, limit=15):
        """Extract article links and basic info from a category page"""
        articles = []
        
        # Find article elements using multiple selectors
        article_selectors = [
            'article',
            'li[class*="article"]',
            'li[class*="promo"]',
            'div[class*="story"]',
            'div[class*="card"]',
            'div[class*="teaser"]',
            'div[class*="promo"]',
            'div[data-entityid]'
        ]
        
        article_elements = []
        for selector in article_selectors:
            elements = soup.select(selector)
            if elements:
                article_elements = elements
                break
        
        # Fallback: find any divs with article-like classes
        if not article_elements:
            for div in soup.find_all('div'):
                class_list = div.get('class', [])
                class_str = ' '.join(class_list) if class_list else ''
                if any(keyword in class_str.lower() for keyword in ['story', 'article', 'promo', 'card']):
                    article_elements.append(div)
        
        # Fallback: find list items
        if not article_elements:
            for li in soup.find_all('li'):
                class_list = li.get('class', [])
                class_str = ' '.join(class_list) if class_list else ''
                if any(keyword in class_str.lower() for keyword in ['story', 'article', 'promo']):
                    article_elements.append(li)
        
        for element in article_elements[:limit]:
            try:
                # Extract title
                title = ""
                title_selectors = ['h1', 'h2', 'h3', 'h4', '[class*="title"]', '[class*="heading"]']
                for selector in title_selectors:
                    title_elem = element.find(selector)
                    if title_elem:
                        title = self.clean_text(title_elem.get_text(strip=True))
                        break
                
                if not title:
                    link_elem = element.find('a')
                    if link_elem:
                        title = self.clean_text(link_elem.get_text(strip=True))
                
                if not title or len(title) < 3:
                    continue
                
                # Extract link
                link = ""
                link_elem = element.find('a')
                if link_elem:
                    link = link_elem.get('href', '')
                    if link and not link.startswith('http'):
                        link = urljoin(self.base_url, link)
                
                # Skip duplicates or invalid links
                if not link or link in self.article_links:
                    continue
                if 'bbc.com' not in link:
                    continue
                
                self.article_links.add(link)
                
                # Extract date from category page
                date = ""
                date_selectors = ['time', '[class*="date"]', '[class*="time"]', '[class*="published"]']
                for selector in date_selectors:
                    date_elem = element.find(selector)
                    if date_elem:
                        date = self.clean_text(date_elem.get_text(strip=True))
                        if date:
                            break
                
                # Extract summary from category page
                summary = ""
                summary_selectors = ['p', '[class*="summary"]', '[class*="description"]', '[class*="excerpt"]']
                for selector in summary_selectors:
                    summary_elem = element.find(selector)
                    if summary_elem:
                        summary = self.clean_text(summary_elem.get_text(strip=True))
                        if summary:
                            break
                
                articles.append({
                    'category': category_name,
                    'category_display': category_display,
                    'title': title,
                    'link': link,
                    'date': date,
                    'summary': summary,
                    'full_content': '',
                    'image_url': ''
                })
                
            except Exception as e:
                continue
        
        return articles

    def scrape_category(self, category):
        """Scrape a specific category page and then fetch full content for each article"""
        category_name = category['name']
        category_display = category['display_name']
        url = category['url']
        
        print(f"\n📂 Scraping: {category_display} ({category_name})")
        
        soup = self.fetch_page(url)
        if not soup:
            print(f"  ❌ Failed to fetch {url}")
            return []
        
        # Extract articles from the category page
        articles = self.extract_articles_from_page(soup, category_name, category_display, limit=10)
        print(f"  📄 Found {len(articles)} articles")
        
        # Fetch full content for each article
        for i, article in enumerate(articles, 1):
            print(f"  📖 [{i}/{len(articles)}] Fetching full content...")
            
            # Get full content and image from the article page
            full_content, img_url = self.extract_full_article_content(article['link'])
            
            article['full_content'] = full_content
            if img_url:
                article['image_url'] = img_url
            
            # Update the image_url if not found on article page
            if not article['image_url']:
                # Try to find image from the category page element
                soup_article = self.fetch_page(article['link'])
                if soup_article:
                    img_elem = soup_article.find('img')
                    if img_elem:
                        img_src = img_elem.get('src', '')
                        if img_src and not img_src.startswith('http'):
                            img_src = urljoin(self.base_url, img_src)
                        article['image_url'] = img_src
            
            # Short delay between article fetches
            time.sleep(0.5)
        
        print(f"  ✅ Completed {category_display}")
        return articles

    def scrape_all_categories(self):
        """Scrape all categories"""
        print("\n" + "="*70)
        print("🎯 BBC AMHARIC FULL CONTENT SCRAPER")
        print("="*70)
        print(f"⏰ Started: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
        print(f"📁 Categories: {len(self.categories)}")
        print("="*70)
        
        total_articles = 0
        for category in self.categories:
            category_articles = self.scrape_category(category)
            self.all_articles.extend(category_articles)
            total_articles += len(category_articles)
            print(f"  📊 Total so far: {total_articles} articles")
            # Delay between categories
            time.sleep(2)
        
        print("\n" + "="*70)
        print(f"✅ SCRAPING COMPLETE!")
        print(f"📊 Total articles scraped: {len(self.all_articles)}")
        print("="*70)
        return self.all_articles

    def save_to_csv(self):
        """Save scraped articles to CSV file"""
        if not self.all_articles:
            print("❌ No articles to save!")
            return None
        
        # Create filename with timestamp
        timestamp = datetime.now().strftime('%Y%m%d_%H%M%S')
        filename = f"bbc_amharic_full_news_{timestamp}.csv"
        
        # Write to CSV
        with open(filename, 'w', newline='', encoding='utf-8-sig') as csvfile:
            fieldnames = ['category', 'category_display', 'title', 'link', 'date', 'summary', 'full_content', 'image_url']
            writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
            
            writer.writeheader()
            for article in self.all_articles:
                writer.writerow(article)
        
        print(f"\n💾 Data saved to: {filename}")
        print(f"📊 Total rows: {len(self.all_articles)}")
        return filename

    def print_summary(self):
        """Print category summary"""
        print("\n" + "="*70)
        print("📊 CATEGORY SUMMARY")
        print("="*70)
        
        stats = {}
        for article in self.all_articles:
            cat = article['category']
            stats[cat] = stats.get(cat, 0) + 1
        
        for category in self.categories:
            name = category['name']
            display = category['display_name']
            count = stats.get(name, 0)
            print(f"  {display:15} : {count:3} articles")
        
        print("="*70)

def main():
    # Create scraper instance
    scraper = BBCAmharicFullScraper()
    
    # Scrape all categories
    scraper.scrape_all_categories()
    
    # Print summary
    scraper.print_summary()
    
    # Save to CSV
    csv_file = scraper.save_to_csv()
    
    # Show sample
    if scraper.all_articles:
        print("\n" + "="*70)
        print("📄 SAMPLE ARTICLE")
        print("="*70)
        sample = scraper.all_articles[0]
        print(f"  Category: {sample['category_display']}")
        print(f"  Title: {sample['title']}")
        print(f"  Date: {sample['date']}")
        print(f"  Link: {sample['link']}")
        print(f"  Summary: {sample['summary'][:150]}..." if sample['summary'] else "  Summary: N/A")
        print(f"  Full Content: {sample['full_content'][:200]}..." if sample['full_content'] else "  Full Content: N/A")
        print(f"  Image URL: {sample['image_url'][:80]}..." if sample['image_url'] else "  Image URL: N/A")
        print("="*70)

if __name__ == "__main__":
    main()
```

```python
import requests
from bs4 import BeautifulSoup
import csv
import time
from datetime import datetime
from urllib.parse import urljoin, urlparse
import re

class BBCAmharicCompleteScraper:
    def __init__(self):
        self.base_url = "https://www.bbc.com/amharic"
        self.categories = [
            {
                'name': 'zena',
                'url': 'https://www.bbc.com/amharic',
                'display_name': 'ዜና'
            },
            {
                'name': 'ethiopia',
                'url': 'https://www.bbc.com/amharic/topics/c7zp57r92v5t',
                'display_name': 'ኢትዮጵያ'
            },
            {
                'name': 'politica',
                'url': 'https://www.bbc.com/amharic/topics/cg7265pj1jvt',
                'display_name': 'ፖለቲካ'
            },
            {
                'name': 'setoch',
                'url': 'https://www.bbc.com/amharic/topics/cnq681w4dq6t',
                'display_name': 'ሴቶች'
            },
            {
                'name': 'tena',
                'url': 'https://www.bbc.com/amharic/topics/cxnyk76p0q2t',
                'display_name': 'ጤና'
            },
            {
                'name': 'sport',
                'url': 'https://www.bbc.com/amharic/topics/cdr56g2x71dt',
                'display_name': 'ስፖርት'
            },
            {
                'name': 'tebeb',
                'url': 'https://www.bbc.com/amharic/topics/cr50yvqzzwpt',
                'display_name': 'ጥበብ'
            },
            {
                'name': 'technology',
                'url': 'https://www.bbc.com/amharic/topics/c06gq8wx467t',
                'display_name': 'ቴክኖሎጂ'
            },
            {
                'name': 'metane_haet',
                'url': 'https://www.bbc.com/amharic/topics/cnq6815jj3xt',
                'display_name': 'ምጣኔ ሃብት'
            }
        ]
        self.headers = {
            'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36',
            'Accept-Language': 'am,en-US;q=0.9,en;q=0.8',
        }
        self.all_articles = []
        self.article_links = set()
        self.total_pages_scraped = 0

    def fetch_page(self, url):
        """Fetch page content with error handling and retry"""
        try:
            response = requests.get(url, headers=self.headers, timeout=15)
            response.raise_for_status()
            return BeautifulSoup(response.content, 'html.parser')
        except requests.RequestException as e:
            print(f"    ❌ Error: {e}")
            return None

    def clean_text(self, text):
        """Clean text by removing extra whitespace and newlines"""
        if not text:
            return ""
        return ' '.join(text.split())

    def extract_full_article_content(self, article_url):
        """Visit the article page and extract the full content"""
        soup = self.fetch_page(article_url)
        if not soup:
            return "", ""
        
        # Extract full content - try multiple selectors
        content = ""
        content_selectors = [
            'div[class*="story-body"]',
            'div[class*="article-body"]',
            'div[class*="content"]',
            'div[class*="story"]',
            'article',
            'main'
        ]
        
        for selector in content_selectors:
            content_elem = soup.select(selector)
            if content_elem:
                paragraphs = content_elem[0].find_all('p')
                if paragraphs:
                    content = ' '.join([self.clean_text(p.get_text(strip=True)) for p in paragraphs])
                    if content:
                        break
        
        # If no content found, try getting all text from article
        if not content:
            article_main = soup.find('article') or soup.find('main')
            if article_main:
                paragraphs = article_main.find_all('p')
                if paragraphs:
                    content = ' '.join([self.clean_text(p.get_text(strip=True)) for p in paragraphs])
        
        # Extract image
        img_url = ""
        img_elem = soup.find('img')
        if img_elem:
            img_url = img_elem.get('src', '')
            if img_url and not img_url.startswith('http'):
                img_url = urljoin(self.base_url, img_url)
        
        return content, img_url

    def extract_articles_from_page(self, soup):
        """Extract all articles from a page"""
        articles = []
        
        # Find article elements
        article_selectors = [
            'article',
            'li[class*="article"]',
            'li[class*="promo"]',
            'div[class*="story"]',
            'div[class*="card"]',
            'div[class*="teaser"]',
            'div[class*="promo"]',
            'div[data-entityid]'
        ]
        
        article_elements = []
        for selector in article_selectors:
            elements = soup.select(selector)
            if elements:
                article_elements = elements
                break
        
        # Fallback: find any divs with article-like classes
        if not article_elements:
            for div in soup.find_all('div'):
                class_list = div.get('class', [])
                class_str = ' '.join(class_list) if class_list else ''
                if any(keyword in class_str.lower() for keyword in ['story', 'article', 'promo', 'card']):
                    article_elements.append(div)
        
        # Fallback: find list items
        if not article_elements:
            for li in soup.find_all('li'):
                class_list = li.get('class', [])
                class_str = ' '.join(class_list) if class_list else ''
                if any(keyword in class_str.lower() for keyword in ['story', 'article', 'promo']):
                    article_elements.append(li)
        
        for element in article_elements:
            try:
                # Extract title
                title = ""
                title_selectors = ['h1', 'h2', 'h3', 'h4', '[class*="title"]', '[class*="heading"]']
                for selector in title_selectors:
                    title_elem = element.find(selector)
                    if title_elem:
                        title = self.clean_text(title_elem.get_text(strip=True))
                        break
                
                if not title:
                    link_elem = element.find('a')
                    if link_elem:
                        title = self.clean_text(link_elem.get_text(strip=True))
                
                if not title or len(title) < 3:
                    continue
                
                # Extract link
                link = ""
                link_elem = element.find('a')
                if link_elem:
                    link = link_elem.get('href', '')
                    if link and not link.startswith('http'):
                        link = urljoin(self.base_url, link)
                
                # Skip duplicates or invalid links
                if not link or link in self.article_links:
                    continue
                if 'bbc.com' not in link:
                    continue
                
                self.article_links.add(link)
                
                # Extract date
                date = ""
                date_selectors = ['time', '[class*="date"]', '[class*="time"]', '[class*="published"]']
                for selector in date_selectors:
                    date_elem = element.find(selector)
                    if date_elem:
                        date = self.clean_text(date_elem.get_text(strip=True))
                        if date:
                            break
                
                # Extract summary
                summary = ""
                summary_selectors = ['p', '[class*="summary"]', '[class*="description"]', '[class*="excerpt"]']
                for selector in summary_selectors:
                    summary_elem = element.find(selector)
                    if summary_elem:
                        summary = self.clean_text(summary_elem.get_text(strip=True))
                        if summary:
                            break
                
                articles.append({
                    'title': title,
                    'link': link,
                    'date': date,
                    'summary': summary
                })
                
            except Exception as e:
                continue
        
        return articles

    def get_pagination_urls(self, soup, base_url):
        """Get all pagination URLs from a category page"""
        pagination_urls = []
        
        # Find pagination links
        pagination_selectors = [
            'nav[class*="pagination"]',
            'div[class*="pagination"]',
            'ul[class*="pagination"]',
            'ol[class*="pagination"]',
            'a[class*="pagination"]'
        ]
        
        for selector in pagination_selectors:
            pagination_elem = soup.select(selector)
            if pagination_elem:
                # Find all page links
                for a in pagination_elem[0].find_all('a', href=True):
                    href = a.get('href')
                    if href:
                        if href.startswith('?') or href.startswith('&'):
                            href = base_url + href
                        elif not href.startswith('http'):
                            href = urljoin(self.base_url, href)
                        pagination_urls.append(href)
                break
        
        # If no pagination found, try finding links with page numbers
        if not pagination_urls:
            for a in soup.find_all('a', href=True):
                href = a.get('href')
                text = a.get_text(strip=True)
                # Look for page numbers
                if text and text.isdigit():
                    if href.startswith('?') or href.startswith('&'):
                        href = base_url + href
                    elif not href.startswith('http'):
                        href = urljoin(self.base_url, href)
                    pagination_urls.append(href)
        
        # Remove duplicates and clean
        pagination_urls = list(set(pagination_urls))
        
        # Keep only URLs that are pagination (contain page numbers or offset)
        clean_urls = []
        for url in pagination_urls:
            if 'page=' in url or 'p=' in url or 'offset=' in url or 'start=' in url:
                clean_urls.append(url)
            # For BBC, look for page numbers in the URL
            elif any(char.isdigit() for char in url.split('/')[-1]):
                clean_urls.append(url)
        
        # Sort to get page 2, 3, 4...
        clean_urls = sorted(clean_urls)
        
        return clean_urls

    def get_last_page_number(self, soup, base_url):
        """Get the last page number from pagination"""
        max_page = 1
        
        # Find pagination
        pagination_selectors = [
            'nav[class*="pagination"]',
            'div[class*="pagination"]',
            'ul[class*="pagination"]'
        ]
        
        for selector in pagination_selectors:
            pagination = soup.select(selector)
            if pagination:
                for a in pagination[0].find_all('a', href=True):
                    text = a.get_text(strip=True)
                    if text and text.isdigit():
                        page_num = int(text)
                        if page_num > max_page:
                            max_page = page_num
        
        # If no pagination found, check for last page link
        if max_page == 1:
            for a in soup.find_all('a', href=True):
                text = a.get_text(strip=True)
                if text and text.isdigit() and int(text) > max_page:
                    max_page = int(text)
        
        return max_page

    def scrape_category_pages(self, category):
        """Scrape all pages of a category"""
        category_name = category['name']
        category_display = category['display_name']
        base_url = category['url']
        
        print(f"\n📂 {category_display} ({category_name})")
        
        # Get first page
        soup = self.fetch_page(base_url)
        if not soup:
            print(f"  ❌ Failed to fetch {base_url}")
            return []
        
        # Get last page number
        last_page = self.get_last_page_number(soup, base_url)
        print(f"  📄 Found {last_page} pages")
        
        all_category_articles = []
        
        # Scrape all pages
        for page_num in range(1, last_page + 1):
            if page_num == 1:
                current_url = base_url
            else:
                # Try different pagination formats
                current_url = f"{base_url}?page={page_num}"
                if '?page=' in base_url:
                    current_url = f"{base_url}&page={page_num}"
            
            print(f"  📖 Page {page_num}/{last_page}: {current_url[:80]}...")
            
            # Fetch the page
            page_soup = self.fetch_page(current_url)
            if not page_soup:
                print(f"    ⚠️ Failed to fetch page {page_num}, trying next...")
                continue
            
            # Extract articles from this page
            page_articles = self.extract_articles_from_page(page_soup)
            print(f"    📝 Found {len(page_articles)} articles")
            
            # Add category info
            for article in page_articles:
                article['category'] = category_name
                article['category_display'] = category_display
                
            all_category_articles.extend(page_articles)
            
            # Delay between pages
            time.sleep(1)
        
        print(f"  ✅ Total: {len(all_category_articles)} articles from {last_page} pages")
        return all_category_articles

    def scrape_all_categories(self):
        """Scrape all categories including all pages"""
        print("\n" + "="*80)
        print("🎯 BBC AMHARIC COMPLETE SCRAPER - ALL ARTICLES ALL PAGES")
        print("="*80)
        print(f"⏰ Started: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
        print(f"📁 Categories: {len(self.categories)}")
        print("="*80)
        
        total_articles = 0
        
        for category in self.categories:
            # Scrape all pages of this category
            category_articles = self.scrape_category_pages(category)
            
            # Fetch full content for each article
            print(f"  📖 Fetching full content for {len(category_articles)} articles...")
            for i, article in enumerate(category_articles, 1):
                print(f"    [{i}/{len(category_articles)}] {article['title'][:50]}...")
                
                full_content, img_url = self.extract_full_article_content(article['link'])
                article['full_content'] = full_content
                article['image_url'] = img_url
                
                self.all_articles.append(article)
                
                # Delay between articles
                time.sleep(0.3)
            
            total_articles += len(category_articles)
            print(f"  ✅ {category['display_name']}: {len(category_articles)} articles")
            print(f"  📊 Total so far: {total_articles} articles")
            
            # Delay between categories
            time.sleep(2)
        
        print("\n" + "="*80)
        print(f"✅ SCRAPING COMPLETE!")
        print(f"📊 Total articles scraped: {len(self.all_articles)}")
        print(f"📊 Total unique links: {len(self.article_links)}")
        print("="*80)
        return self.all_articles

    def save_to_csv(self):
        """Save scraped articles to CSV file"""
        if not self.all_articles:
            print("❌ No articles to save!")
            return None
        
        timestamp = datetime.now().strftime('%Y%m%d_%H%M%S')
        filename = f"bbc_amharic_complete_{timestamp}.csv"
        
        with open(filename, 'w', newline='', encoding='utf-8-sig') as csvfile:
            fieldnames = ['category', 'category_display', 'title', 'link', 'date', 'summary', 'full_content', 'image_url']
            writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
            
            writer.writeheader()
            for article in self.all_articles:
                # Ensure all fields exist
                for field in fieldnames:
                    if field not in article:
                        article[field] = ''
                writer.writerow(article)
        
        print(f"\n💾 Data saved to: {filename}")
        print(f"📊 Total rows: {len(self.all_articles)}")
        return filename

    def print_summary(self):
        """Print detailed summary"""
        print("\n" + "="*80)
        print("📊 SCRAPING SUMMARY")
        print("="*80)
        
        stats = {}
        for article in self.all_articles:
            cat = article['category']
            stats[cat] = stats.get(cat, 0) + 1
        
        total_with_content = sum(1 for a in self.all_articles if a.get('full_content'))
        total_with_images = sum(1 for a in self.all_articles if a.get('image_url'))
        
        print("\n📁 ARTICLES PER CATEGORY:")
        for category in self.categories:
            name = category['name']
            display = category['display_name']
            count = stats.get(name, 0)
            print(f"  {display:15} : {count:4} articles")
        
        print("\n📊 TOTALS:")
        print(f"  Total articles: {len(self.all_articles)}")
        print(f"  With full content: {total_with_content}")
        print(f"  With images: {total_with_images}")
        print("="*80)

def main():
    # Create scraper instance
    scraper = BBCAmharicCompleteScraper()
    
    # Scrape all categories (all pages)
    scraper.scrape_all_categories()
    
    # Print summary
    scraper.print_summary()
    
    # Save to CSV
    csv_file = scraper.save_to_csv()
    
    # Show samples
    if scraper.all_articles:
        print("\n" + "="*80)
        print("📄 SAMPLE ARTICLES (First 3)")
        print("="*80)
        
        for i, article in enumerate(scraper.all_articles[:3], 1):
            print(f"\n{i}. Category: {article.get('category_display', 'N/A')}")
            print(f"   Title: {article.get('title', 'N/A')[:80]}...")
            print(f"   Date: {article.get('date', 'N/A')}")
            print(f"   Link: {article.get('link', 'N/A')[:60]}...")
            print(f"   Content: {article.get('full_content', 'N/A')[:150]}...")
            print("-"*60)

        

if __name__ == "__main__":
    main()
```

```python
import requests
from bs4 import BeautifulSoup
import csv
import time
from datetime import datetime
from urllib.parse import urljoin, urlparse
import re

class BBCAmharicCompleteScraper:
    def __init__(self):
        self.base_url = "https://www.bbc.com/amharic"
        self.categories = [
            {
                'name': 'zena',
                'url': 'https://www.bbc.com/amharic',
                'display_name': 'ዜና'
            },
            {
                'name': 'ethiopia',
                'url': 'https://www.bbc.com/amharic/topics/c7zp57r92v5t',
                'display_name': 'ኢትዮጵያ'
            },
            {
                'name': 'politica',
                'url': 'https://www.bbc.com/amharic/topics/cg7265pj1jvt',
                'display_name': 'ፖለቲካ'
            },
            {
                'name': 'setoch',
                'url': 'https://www.bbc.com/amharic/topics/cnq681w4dq6t',
                'display_name': 'ሴቶች'
            },
            {
                'name': 'tena',
                'url': 'https://www.bbc.com/amharic/topics/cxnyk76p0q2t',
                'display_name': 'ጤና'
            },
            {
                'name': 'sport',
                'url': 'https://www.bbc.com/amharic/topics/cdr56g2x71dt',
                'display_name': 'ስፖርት'
            },
            {
                'name': 'tebeb',
                'url': 'https://www.bbc.com/amharic/topics/cr50yvqzzwpt',
                'display_name': 'ጥበብ'
            },
            {
                'name': 'technology',
                'url': 'https://www.bbc.com/amharic/topics/c06gq8wx467t',
                'display_name': 'ቴክኖሎጂ'
            },
            {
                'name': 'metane_haet',
                'url': 'https://www.bbc.com/amharic/topics/cnq6815jj3xt',
                'display_name': 'ምጣኔ ሃብት'
            }
        ]
        self.headers = {
            'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36',
            'Accept-Language': 'am,en-US;q=0.9,en;q=0.8',
        }
        self.all_articles = []
        self.article_links = set()
        self.total_pages_scraped = 0

    def fetch_page(self, url):
        """Fetch page content with error handling and retry"""
        try:
            response = requests.get(url, headers=self.headers, timeout=15)
            response.raise_for_status()
            return BeautifulSoup(response.content, 'html.parser')
        except requests.RequestException as e:
            print(f"    ❌ Error: {e}")
            return None

    def clean_text(self, text):
        """Clean text by removing extra whitespace and newlines"""
        if not text:
            return ""
        return ' '.join(text.split())

    def extract_full_article_content(self, article_url):
        """Visit the article page and extract the full content"""
        soup = self.fetch_page(article_url)
        if not soup:
            return "", ""
        
        # Extract full content - try multiple selectors
        content = ""
        content_selectors = [
            'div[class*="story-body"]',
            'div[class*="article-body"]',
            'div[class*="content"]',
            'div[class*="story"]',
            'article',
            'main'
        ]
        
        for selector in content_selectors:
            content_elem = soup.select(selector)
            if content_elem:
                paragraphs = content_elem[0].find_all('p')
                if paragraphs:
                    content = ' '.join([self.clean_text(p.get_text(strip=True)) for p in paragraphs])
                    if content:
                        break
        
        # If no content found, try getting all text from article
        if not content:
            article_main = soup.find('article') or soup.find('main')
            if article_main:
                paragraphs = article_main.find_all('p')
                if paragraphs:
                    content = ' '.join([self.clean_text(p.get_text(strip=True)) for p in paragraphs])
        
        # Extract image
        img_url = ""
        img_elem = soup.find('img')
        if img_elem:
            img_url = img_elem.get('src', '')
            if img_url and not img_url.startswith('http'):
                img_url = urljoin(self.base_url, img_url)
        
        return content, img_url

    def extract_articles_from_page(self, soup):
        """Extract all articles from a page"""
        articles = []
        
        # Find article elements
        article_selectors = [
            'article',
            'li[class*="article"]',
            'li[class*="promo"]',
            'div[class*="story"]',
            'div[class*="card"]',
            'div[class*="teaser"]',
            'div[class*="promo"]',
            'div[data-entityid]'
        ]
        
        article_elements = []
        for selector in article_selectors:
            elements = soup.select(selector)
            if elements:
                article_elements = elements
                break
        
        # Fallback: find any divs with article-like classes
        if not article_elements:
            for div in soup.find_all('div'):
                class_list = div.get('class', [])
                class_str = ' '.join(class_list) if class_list else ''
                if any(keyword in class_str.lower() for keyword in ['story', 'article', 'promo', 'card']):
                    article_elements.append(div)
        
        # Fallback: find list items
        if not article_elements:
            for li in soup.find_all('li'):
                class_list = li.get('class', [])
                class_str = ' '.join(class_list) if class_list else ''
                if any(keyword in class_str.lower() for keyword in ['story', 'article', 'promo']):
                    article_elements.append(li)
        
        for element in article_elements:
            try:
                # Extract title
                title = ""
                title_selectors = ['h1', 'h2', 'h3', 'h4', '[class*="title"]', '[class*="heading"]']
                for selector in title_selectors:
                    title_elem = element.find(selector)
                    if title_elem:
                        title = self.clean_text(title_elem.get_text(strip=True))
                        break
                
                if not title:
                    link_elem = element.find('a')
                    if link_elem:
                        title = self.clean_text(link_elem.get_text(strip=True))
                
                if not title or len(title) < 3:
                    continue
                
                # Extract link
                link = ""
                link_elem = element.find('a')
                if link_elem:
                    link = link_elem.get('href', '')
                    if link and not link.startswith('http'):
                        link = urljoin(self.base_url, link)
                
                # Skip duplicates or invalid links
                if not link or link in self.article_links:
                    continue
                if 'bbc.com' not in link:
                    continue
                
                self.article_links.add(link)
                
                # Extract date
                date = ""
                date_selectors = ['time', '[class*="date"]', '[class*="time"]', '[class*="published"]']
                for selector in date_selectors:
                    date_elem = element.find(selector)
                    if date_elem:
                        date = self.clean_text(date_elem.get_text(strip=True))
                        if date:
                            break
                
                # Extract summary
                summary = ""
                summary_selectors = ['p', '[class*="summary"]', '[class*="description"]', '[class*="excerpt"]']
                for selector in summary_selectors:
                    summary_elem = element.find(selector)
                    if summary_elem:
                        summary = self.clean_text(summary_elem.get_text(strip=True))
                        if summary:
                            break
                
                articles.append({
                    'title': title,
                    'link': link,
                    'date': date,
                    'summary': summary
                })
                
            except Exception as e:
                continue
        
        return articles

    def get_pagination_urls(self, soup, base_url):
        """Get all pagination URLs from a category page"""
        pagination_urls = []
        
        # Find pagination links
        pagination_selectors = [
            'nav[class*="pagination"]',
            'div[class*="pagination"]',
            'ul[class*="pagination"]',
            'ol[class*="pagination"]',
            'a[class*="pagination"]'
        ]
        
        for selector in pagination_selectors:
            pagination_elem = soup.select(selector)
            if pagination_elem:
                # Find all page links
                for a in pagination_elem[0].find_all('a', href=True):
                    href = a.get('href')
                    if href:
                        if href.startswith('?') or href.startswith('&'):
                            href = base_url + href
                        elif not href.startswith('http'):
                            href = urljoin(self.base_url, href)
                        pagination_urls.append(href)
                break
        
        # If no pagination found, try finding links with page numbers
        if not pagination_urls:
            for a in soup.find_all('a', href=True):
                href = a.get('href')
                text = a.get_text(strip=True)
                # Look for page numbers
                if text and text.isdigit():
                    if href.startswith('?') or href.startswith('&'):
                        href = base_url + href
                    elif not href.startswith('http'):
                        href = urljoin(self.base_url, href)
                    pagination_urls.append(href)
        
        # Remove duplicates and clean
        pagination_urls = list(set(pagination_urls))
        
        # Keep only URLs that are pagination (contain page numbers or offset)
        clean_urls = []
        for url in pagination_urls:
            if 'page=' in url or 'p=' in url or 'offset=' in url or 'start=' in url:
                clean_urls.append(url)
            # For BBC, look for page numbers in the URL
            elif any(char.isdigit() for char in url.split('/')[-1]):
                clean_urls.append(url)
        
        # Sort to get page 2, 3, 4...
        clean_urls = sorted(clean_urls)
        
        return clean_urls

    def get_last_page_number(self, soup, base_url):
        """Get the last page number from pagination"""
        max_page = 1
        
        # Find pagination
        pagination_selectors = [
            'nav[class*="pagination"]',
            'div[class*="pagination"]',
            'ul[class*="pagination"]'
        ]
        
        for selector in pagination_selectors:
            pagination = soup.select(selector)
            if pagination:
                for a in pagination[0].find_all('a', href=True):
                    text = a.get_text(strip=True)
                    if text and text.isdigit():
                        page_num = int(text)
                        if page_num > max_page:
                            max_page = page_num
        
        # If no pagination found, check for last page link
        if max_page == 1:
            for a in soup.find_all('a', href=True):
                text = a.get_text(strip=True)
                if text and text.isdigit() and int(text) > max_page:
                    max_page = int(text)
        
        return max_page

    def scrape_category_pages(self, category):
        """Scrape all pages of a category"""
        category_name = category['name']
        category_display = category['display_name']
        base_url = category['url']
        
        print(f"\n📂 {category_display} ({category_name})")
        
        # Get first page
        soup = self.fetch_page(base_url)
        if not soup:
            print(f"  ❌ Failed to fetch {base_url}")
            return []
        
        # Get last page number
        last_page = self.get_last_page_number(soup, base_url)
        print(f"  📄 Found {last_page} pages")
        
        all_category_articles = []
        
        # Scrape all pages
        for page_num in range(1, last_page + 1):
            if page_num == 1:
                current_url = base_url
            else:
                # Try different pagination formats
                current_url = f"{base_url}?page={page_num}"
                if '?page=' in base_url:
                    current_url = f"{base_url}&page={page_num}"
            
            print(f"  📖 Page {page_num}/{last_page}: {current_url[:80]}...")
            
            # Fetch the page
            page_soup = self.fetch_page(current_url)
            if not page_soup:
                print(f"    ⚠️ Failed to fetch page {page_num}, trying next...")
                continue
            
            # Extract articles from this page
            page_articles = self.extract_articles_from_page(page_soup)
            print(f"    📝 Found {len(page_articles)} articles")
            
            # Add category info
            for article in page_articles:
                article['category'] = category_name
                article['category_display'] = category_display
                
            all_category_articles.extend(page_articles)
            
            # Delay between pages
            time.sleep(1)
        
        print(f"  ✅ Total: {len(all_category_articles)} articles from {last_page} pages")
        return all_category_articles

    def scrape_all_categories(self):
        """Scrape all categories including all pages"""
        print("\n" + "="*80)
        print("🎯 BBC AMHARIC COMPLETE SCRAPER - ALL ARTICLES ALL PAGES")
        print("="*80)
        print(f"⏰ Started: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
        print(f"📁 Categories: {len(self.categories)}")
        print("="*80)
        
        total_articles = 0
        
        for category in self.categories:
            # Scrape all pages of this category
            category_articles = self.scrape_category_pages(category)
            
            # Fetch full content for each article
            print(f"  📖 Fetching full content for {len(category_articles)} articles...")
            for i, article in enumerate(category_articles, 1):
                print(f"    [{i}/{len(category_articles)}] {article['title'][:50]}...")
                
                full_content, img_url = self.extract_full_article_content(article['link'])
                article['full_content'] = full_content
                article['image_url'] = img_url
                
                self.all_articles.append(article)
                
                # Delay between articles
                time.sleep(0.3)
            
            total_articles += len(category_articles)
            print(f"  ✅ {category['display_name']}: {len(category_articles)} articles")
            print(f"  📊 Total so far: {total_articles} articles")
            
            # Delay between categories
            time.sleep(2)
        
        print("\n" + "="*80)
        print(f"✅ SCRAPING COMPLETE!")
        print(f"📊 Total articles scraped: {len(self.all_articles)}")
        print(f"📊 Total unique links: {len(self.article_links)}")
        print("="*80)
        return self.all_articles

    def save_to_csv(self):
        """Save scraped articles to CSV file"""
        if not self.all_articles:
            print("❌ No articles to save!")
            return None
        
        timestamp = datetime.now().strftime('%Y%m%d_%H%M%S')
        filename = f"bbc_amharic_complete_{timestamp}.csv"
        
        with open(filename, 'w', newline='', encoding='utf-8-sig') as csvfile:
            fieldnames = ['category', 'category_display', 'title', 'link', 'date', 'summary', 'full_content', 'image_url']
            writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
            
            writer.writeheader()
            for article in self.all_articles:
                # Ensure all fields exist
                for field in fieldnames:
                    if field not in article:
                        article[field] = ''
                writer.writerow(article)
        
        print(f"\n💾 Data saved to: {filename}")
        print(f"📊 Total rows: {len(self.all_articles)}")
        return filename

    def print_summary(self):
        """Print detailed summary"""
        print("\n" + "="*80)
        print("📊 SCRAPING SUMMARY")
        print("="*80)
        
        stats = {}
        for article in self.all_articles:
            cat = article['category']
            stats[cat] = stats.get(cat, 0) + 1
        
        total_with_content = sum(1 for a in self.all_articles if a.get('full_content'))
        total_with_images = sum(1 for a in self.all_articles if a.get('image_url'))
        
        print("\n📁 ARTICLES PER CATEGORY:")
        for category in self.categories:
            name = category['name']
            display = category['display_name']
            count = stats.get(name, 0)
            print(f"  {display:15} : {count:4} articles")
        
        print("\n📊 TOTALS:")
        print(f"  Total articles: {len(self.all_articles)}")
        print(f"  With full content: {total_with_content}")
        print(f"  With images: {total_with_images}")
        print("="*80)

def main():
    # Create scraper instance
    scraper = BBCAmharicCompleteScraper()
    
    # Scrape all categories (all pages)
    scraper.scrape_all_categories()
    
    # Print summary
    scraper.print_summary()
    
    # Save to CSV
    csv_file = scraper.save_to_csv()
    
    # Show samples
    if scraper.all_articles:
        print("\n" + "="*80)
        print("📄 SAMPLE ARTICLES (First 3)")
        print("="*80)
        
        for i, article in enumerate(scraper.all_articles[:3], 1):
            print(f"\n{i}. Category: {article.get('category_display', 'N/A')}")
            print(f"   Title: {article.get('title', 'N/A')[:80]}...")
            print(f"   Date: {article.get('date', 'N/A')}")
            print(f"   Link: {article.get('link', 'N/A')[:60]}...")
            print(f"   Content: {article.get('full_content', 'N/A')[:150]}...")
            print("-"*60)

if __name__ == "__main__":
    main()
```

```python
import pandas as pd
import dash
from dash import dcc, html, Input, Output
import plotly.express as px
import re
from datetime import datetime
import os
import numpy as np

# Load data
def load_data():
    filename = "bbc_amharic_complete_20260624_024105.csv"
    
    if not os.path.exists(filename):
        print(f"File '{filename}' not found!")
        return None
    
    df = pd.read_csv(filename, encoding='utf-8-sig')
    
    # Clean unwanted text from content
    patterns = [
        r'የፎቶው ባለመብት.*?$',
        r'Photo credit.*?$',
        r'Credit:.*?$',
        r'Image credit.*?$',
        r'Getty Images.*?$',
        r'AP Photo.*?$',
        r'Reuters.*?$',
        r'AFP.*?$',
    ]
    
    for col in ['full_content', 'summary', 'title']:
        if col in df.columns:
            for pattern in patterns:
                df[col] = df[col].str.replace(pattern, '', regex=True, flags=re.IGNORECASE)
            df[col] = df[col].str.replace(r'\s+', ' ', regex=True).str.strip()
    
    # Parse date
    if 'date' in df.columns:
        try:
            df['date_parsed'] = pd.to_datetime(df['date'], errors='coerce')
        except:
            df['date_parsed'] = datetime.now()
    else:
        df['date_parsed'] = datetime.now()
    
    # Word count
    if 'full_content' in df.columns:
        df['word_count'] = df['full_content'].str.split().str.len().fillna(0).astype(int)
    else:
        df['word_count'] = 0
    
    # Category display names
    category_names = {
        'zena': 'ዜና', 
        'ethiopia': 'ኢትዮጵያ', 
        'politica': 'ፖለቲካ',
        'setoch': 'ሴቶች', 
        'tena': 'ጤና', 
        'sport': 'ስፖርት',
        'tebeb': 'ጥበብ', 
        'technology': 'ቴክኖሎጂ', 
        'metane_haet': 'ምጣኔ ሃብት'
    }
    
    if 'category' in df.columns:
        df['category_display'] = df['category'].map(category_names).fillna(df['category'])
    else:
        df['category_display'] = 'Unknown'
    
    return df

def gregorian_to_ethiopian(date_obj):
    """
    Convert Gregorian date to Ethiopian calendar
    Ethiopian calendar is 7-8 years behind Gregorian
    Ethiopian New Year is on September 11 (or 12 in leap years)
    """
    if pd.isna(date_obj):
        return "", "", ""
    
    year = date_obj.year
    month = date_obj.month
    day = date_obj.day
    
    # Ethiopian months
    ethiopian_months = [
        'መስከረም', 'ጥቅምት', 'ህዳር', 'ታህሳስ',
        'ጥር', 'የካቲት', 'መጋቢት', 'ሚያዝያ',
        'ግንቦት', 'ሰኔ', 'ሐምሌ', 'ነሐሴ',
        'ጳጉሜ'  # 13th month (5 or 6 days)
    ]
    
    # Ethiopian New Year is September 11 (or 12 in leap years)
    # For simplicity, we'll use approximate conversion
    
    # Calculate Ethiopian year (7-8 years behind)
    eth_year = year - 8
    
    # Calculate day of year (1 = Jan 1)
    start_of_year = datetime(year, 1, 1)
    day_of_year = (date_obj - start_of_year).days + 1
    
    # Determine if it's before or after Ethiopian New Year (Sept 11)
    if month < 9 or (month == 9 and day < 11):
        # Before Ethiopian New Year
        eth_year = year - 8
        # Days before Sept 11
        days_before = (datetime(year, 9, 11) - date_obj).days
        if days_before < 0:
            days_before = 0
        # Calculate Ethiopian month and day
        if day_of_year <= 164:  # Up to June 14
            # First 10 months
            eth_month_index = (day_of_year - 1) // 30
            eth_day = (day_of_year - 1) % 30 + 1
        else:
            # Remaining months
            remaining_days = day_of_year - 164
            if remaining_days <= 30:
                eth_month_index = 10  # ሐምሌ
                eth_day = remaining_days
            elif remaining_days <= 60:
                eth_month_index = 11  # ነሐሴ
                eth_day = remaining_days - 30
            else:
                eth_month_index = 12  # ጳጉሜ
                eth_day = remaining_days - 60
    else:
        # After or on Ethiopian New Year
        eth_year = year - 7
        # Days after Sept 11
        days_after = (date_obj - datetime(year, 9, 11)).days + 1
        if days_after <= 30:
            eth_month_index = 0  # መስከረም
            eth_day = days_after
        elif days_after <= 60:
            eth_month_index = 1  # ጥቅምት
            eth_day = days_after - 30
        elif days_after <= 90:
            eth_month_index = 2  # ህዳር
            eth_day = days_after - 60
        elif days_after <= 120:
            eth_month_index = 3  # ታህሳስ
            eth_day = days_after - 90
        elif days_after <= 150:
            eth_month_index = 4  # ጥር
            eth_day = days_after - 120
        elif days_after <= 180:
            eth_month_index = 5  # የካቲት
            eth_day = days_after - 150
        elif days_after <= 210:
            eth_month_index = 6  # መጋቢት
            eth_day = days_after - 180
        elif days_after <= 240:
            eth_month_index = 7  # ሚያዝያ
            eth_day = days_after - 210
        elif days_after <= 270:
            eth_month_index = 8  # ግንቦት
            eth_day = days_after - 240
        elif days_after <= 300:
            eth_month_index = 9  # ሰኔ
            eth_day = days_after - 270
        elif days_after <= 330:
            eth_month_index = 10  # ሐምሌ
            eth_day = days_after - 300
        elif days_after <= 360:
            eth_month_index = 11  # ነሐሴ
            eth_day = days_after - 330
        else:
            eth_month_index = 12  # ጳጉሜ
            eth_day = days_after - 360
    
    eth_month = ethiopian_months[eth_month_index] if eth_month_index < len(ethiopian_months) else 'ጳጉሜ'
    
    # Ensure day is within valid range
    if eth_day > 30:
        eth_day = 30
    if eth_day < 1:
        eth_day = 1
    
    return eth_month, eth_day, eth_year

def date_to_amharic(date_obj):
    """Convert date to Amharic format with Ethiopian calendar"""
    if pd.isna(date_obj):
        return ""
    
    if isinstance(date_obj, str):
        try:
            date_obj = pd.to_datetime(date_obj)
        except:
            return date_obj
    
    eth_month, eth_day, eth_year = gregorian_to_ethiopian(date_obj)
    
    # Convert numbers to Amharic numerals
    amharic_numerals = {
        '0': '፩', '1': '፩', '2': '፪', '3': '፫', '4': '፬',
        '5': '፭', '6': '፮', '7': '፯', '8': '፰', '9': '፱',
        '10': '፲', '11': '፲፩', '12': '፲፪', '13': '፲፫', '14': '፲፬',
        '15': '፲፭', '16': '፲፮', '17': '፲፯', '18': '፲፰', '19': '፲፱',
        '20': '፳', '21': '፳፩', '22': '፳፪', '23': '፳፫', '24': '፳፬',
        '25': '፳፭', '26': '፳፮', '27': '፳፯', '28': '፳፰', '29': '፳፱',
        '30': '፴'
    }
    
    day_str = str(eth_day)
    day_amharic = amharic_numerals.get(day_str, day_str)
    
    year_str = str(eth_year)
    year_amharic = ''
    for digit in year_str:
        if digit in amharic_numerals:
            year_amharic += amharic_numerals[digit]
        else:
            year_amharic += digit
    
    return f"{eth_month} {day_amharic}, {year_amharic}"

# Load data
df = load_data()
if df is None:
    print("No data loaded. Exiting...")
    exit()

# Check if dataframe is empty
if df.empty:
    print("Dataframe is empty!")
    exit()

# Clean date - remove any invalid dates
df['date_parsed'] = pd.to_datetime(df['date_parsed'], errors='coerce')
df = df.dropna(subset=['date_parsed'])

# Ensure we have data
if df.empty:
    print("No valid date data!")
    exit()

# Get min and max dates
min_date = df['date_parsed'].min()
max_date = df['date_parsed'].max()

# App
app = dash.Dash(__name__)

app.layout = html.Div([

    html.H2("📰 BBC አማርኛ ዜና ዳሽቦርድ", style={'textAlign': 'center', 'color': '#0d47a1'}),

    html.Div([
        # Date Range Filter
        html.Div([
            html.Label("📅 ቀን", style={'fontWeight': 'bold'}),
            dcc.DatePickerRange(
                id="date-range",
                start_date=min_date,
                end_date=max_date,
                display_format='YYYY-MM-DD',
                style={'width': '100%'}
            )
        ], style={'width': '45%', 'display': 'inline-block', 'padding': '10px', 'verticalAlign': 'top'}),

        # Category Filter
        html.Div([
            html.Label("📂 ምድብ", style={'fontWeight': 'bold'}),
            dcc.Dropdown(
                id="category-filter",
                options=[{"label": c, "value": c} for c in df["category"].unique()] if 'category' in df.columns else [],
                multi=True,
                placeholder="ምድብ ይምረጡ",
                style={'width': '100%'}
            )
        ], style={'width': '45%', 'display': 'inline-block', 'padding': '10px', 'verticalAlign': 'top'})
    ], style={'padding': '20px', 'backgroundColor': '#f8f9fa', 'borderRadius': '10px'}),

    # Stats Row
    html.Div([
        html.Div(id="total-articles", style={'display': 'inline-block', 'padding': '20px', 'fontSize': '20px'}),
        html.Div(id="total-categories", style={'display': 'inline-block', 'padding': '20px', 'fontSize': '20px'}),
        html.Div(id="total-words", style={'display': 'inline-block', 'padding': '20px', 'fontSize': '20px'}),
    ], style={'textAlign': 'center', 'padding': '10px'}),

    # Charts
    html.Div([
        dcc.Graph(id="category-chart", style={'width': '48%', 'display': 'inline-block'}),
        dcc.Graph(id="trend-chart", style={'width': '48%', 'display': 'inline-block'})
    ]),

    # Articles Display
    html.Div([
        html.H3("📰 ዜናዎች"),
        html.Div(id="articles-list")
    ], style={'padding': '20px'})
])

@app.callback(
    [
        Output("category-chart", "figure"),
        Output("trend-chart", "figure"),
        Output("articles-list", "children"),
        Output("total-articles", "children"),
        Output("total-categories", "children"),
        Output("total-words", "children")
    ],
    [
        Input("date-range", "start_date"),
        Input("date-range", "end_date"),
        Input("category-filter", "value")
    ]
)
def update_dashboard(start_date, end_date, categories):
    
    dff = df.copy()
    
    # Ensure datetime format
    dff["date_parsed"] = pd.to_datetime(dff["date_parsed"], errors="coerce")
    dff = dff.dropna(subset=['date_parsed'])
    
    # start_date filter
    if start_date is not None and start_date != "":
        start_date = pd.to_datetime(start_date)
        dff = dff[dff["date_parsed"] >= start_date]
    
    # end_date filter
    if end_date is not None and end_date != "":
        end_date = pd.to_datetime(end_date)
        dff = dff[dff["date_parsed"] <= end_date]
    
    # Category filter
    if categories and len(categories) > 0:
        dff = dff[dff["category"].isin(categories)]
    
    # If no data after filtering, create empty results
    if dff.empty:
        fig_pie = px.pie(title="ዜናዎች በምድብ (ምንም ውሂብ የለም)")
        fig_line = px.line(title="ዜናዎች በቀን (ምንም ውሂብ የለም)")
        
        return (
            fig_pie, 
            fig_line, 
            [html.P("ምንም ዜና አልተገኘም", style={'textAlign': 'center', 'padding': '20px'})],
            "📰 ዜናዎች: 0",
            "📂 ምድቦች: 0",
            "📝 ቃላት: 0"
        )
    
    # Stats
    total_articles = f"📰 ዜናዎች: {len(dff)}"
    total_categories = f"📂 ምድቦች: {dff['category'].nunique()}"
    total_words = f"📝 ቃላት: {dff['word_count'].sum():,}" if 'word_count' in dff.columns else "📝 ቃላት: 0"
    
    # Category chart
    if 'category_display' in dff.columns and not dff.empty:
        cat_counts = dff['category_display'].value_counts()
        fig_pie = px.pie(
            values=cat_counts.values,
            names=cat_counts.index,
            title="ዜናዎች በምድብ"
        )
    else:
        fig_pie = px.pie(title="ዜናዎች በምድብ (ምንም ውሂብ የለም)")
    
    # Trend chart
    if 'date_parsed' in dff.columns and not dff.empty:
        daily_counts = dff.groupby(dff['date_parsed'].dt.date).size()
        fig_line = px.line(
            x=daily_counts.index,
            y=daily_counts.values,
            title="ዜናዎች በቀን",
            labels={'x': 'ቀን', 'y': 'ብዛት'}
        )
    else:
        fig_line = px.line(title="ዜናዎች በቀን (ምንም ውሂብ የለም)")
    
    # Articles list
    articles_html = []
    
    # Sort by date (newest first)
    dff = dff.sort_values('date_parsed', ascending=False)
    
    # Colors for categories
    colors = {
        'zena': '#FF6B6B', 'ethiopia': '#4ECDC4', 'politica': '#45B7D1',
        'setoch': '#96CEB4', 'tena': '#FFEAA7', 'sport': '#DDA0DD',
        'tebeb': '#98D8C8', 'technology': '#F7DC6F', 'metane_haet': '#BB8FCE'
    }
    
    for _, row in dff.iterrows():
        # Get content and clean
        content = row.get('full_content', row.get('summary', ''))
        if isinstance(content, str):
            content = re.sub(r'የፎቶው ባለመብት.*?$', '', content, flags=re.IGNORECASE)
            content = re.sub(r'Photo credit.*?$', '', content, flags=re.IGNORECASE)
            content = content.strip()
        else:
            content = str(content)
        
        # Truncate
        if len(content) > 500:
            content = content[:500] + "..."
        
        category = row.get('category', '')
        color = colors.get(category, '#95A5A6')
        cat_display = row.get('category_display', category)
        
        # Convert date to Ethiopian calendar with Amharic
        date_amharic = date_to_amharic(row.get('date_parsed'))
        
        # Get title
        title = row.get('title', 'No Title')
        if pd.isna(title):
            title = 'No Title'
        
        article_html = html.Div([
            html.H4(title, style={'color': '#0d47a1', 'marginBottom': '5px'}),
            html.Div([
                html.Span(cat_display, style={
                    'backgroundColor': color, 
                    'color': 'white', 
                    'padding': '2px 10px', 
                    'borderRadius': '20px',
                    'fontSize': '12px',
                    'marginRight': '10px'
                }),
                html.Span(f"📅 {date_amharic}", style={'marginRight': '10px'}),
                html.Span(f"📊 {row.get('word_count', 0)} ቃላት", style={'marginRight': '10px'}),
                html.A("🔗 አገናኝ", href=row.get('link', '#'), target='_blank', style={'color': '#0d47a1'})
            ], style={'marginBottom': '10px'}),
            html.P(content if content else 'No content available', style={'color': '#333', 'lineHeight': '1.6'}),
            html.Hr()
        ], style={
            'backgroundColor': 'white', 
            'padding': '15px', 
            'borderRadius': '10px',
            'boxShadow': '0 2px 8px rgba(0,0,0,0.1)',
            'marginBottom': '15px',
            'borderLeft': f'5px solid {color}'
        })
        
        articles_html.append(article_html)
    
    return fig_pie, fig_line, articles_html, total_articles, total_categories, total_words

if __name__ == "__main__":
    app.run(debug=True)
```

```python
import pandas as pd
import dash
from dash import dcc, html, Input, Output
import plotly.express as px
import re
from datetime import datetime
import os
import requests
from bs4 import BeautifulSoup
import time

# ==================== SCRAPER ====================
def scrape_bbc_news():
    """Scrape BBC News articles from all categories"""
    
    categories = [
        {'name': 'world', 'url': 'https://www.bbc.com/news', 'display': 'World'},
        {'name': 'us', 'url': 'https://www.bbc.com/news/world/us_and_canada', 'display': 'US & Canada'},
        {'name': 'uk', 'url': 'https://www.bbc.com/news/uk', 'display': 'UK'},
        {'name': 'business', 'url': 'https://www.bbc.com/news/business', 'display': 'Business'},
        {'name': 'tech', 'url': 'https://www.bbc.com/news/technology', 'display': 'Technology'},
        {'name': 'science', 'url': 'https://www.bbc.com/news/science_and_environment', 'display': 'Science'},
        {'name': 'health', 'url': 'https://www.bbc.com/news/health', 'display': 'Health'},
        {'name': 'sport', 'url': 'https://www.bbc.com/sport', 'display': 'Sport'},
        {'name': 'entertainment', 'url': 'https://www.bbc.com/news/entertainment_and_arts', 'display': 'Entertainment'}
    ]
    
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
    }
    
    all_articles = []
    article_links = set()
    
    def clean_text(text):
        if not text:
            return ""
        return ' '.join(text.split())
    
    def fetch_page(url):
        try:
            response = requests.get(url, headers=headers, timeout=15)
            response.raise_for_status()
            return BeautifulSoup(response.content, 'html.parser')
        except Exception as e:
            print(f"Error fetching {url}: {e}")
            return None
    
    def extract_articles_from_page(soup, category_name, category_display, limit=15):
        articles = []
        
        # Find article elements
        article_selectors = [
            'article',
            'li[class*="article"]',
            'li[class*="promo"]',
            'div[class*="story"]',
            'div[class*="card"]',
            'div[class*="teaser"]',
            'div[class*="promo"]'
        ]
        
        article_elements = []
        for selector in article_selectors:
            elements = soup.select(selector)
            if elements:
                article_elements = elements
                break
        
        # Fallback
        if not article_elements:
            for div in soup.find_all('div'):
                class_list = div.get('class', [])
                class_str = ' '.join(class_list) if class_list else ''
                if any(keyword in class_str.lower() for keyword in ['story', 'article', 'promo', 'card']):
                    article_elements.append(div)
        
        for element in article_elements[:limit]:
            try:
                # Extract title
                title = ""
                title_selectors = ['h1', 'h2', 'h3', 'h4', '[class*="title"]', '[class*="heading"]']
                for selector in title_selectors:
                    title_elem = element.find(selector)
                    if title_elem:
                        title = clean_text(title_elem.get_text(strip=True))
                        break
                
                if not title:
                    link_elem = element.find('a')
                    if link_elem:
                        title = clean_text(link_elem.get_text(strip=True))
                
                if not title or len(title) < 3:
                    continue
                
                # Extract link
                link = ""
                link_elem = element.find('a')
                if link_elem:
                    link = link_elem.get('href', '')
                    if link and not link.startswith('http'):
                        link = 'https://www.bbc.com' + link
                
                if not link or link in article_links:
                    continue
                if 'bbc.com' not in link:
                    continue
                
                article_links.add(link)
                
                # Extract date
                date = ""
                date_selectors = ['time', '[class*="date"]', '[class*="time"]', '[class*="published"]']
                for selector in date_selectors:
                    date_elem = element.find(selector)
                    if date_elem:
                        date = clean_text(date_elem.get_text(strip=True))
                        if date:
                            break
                
                # Extract summary
                summary = ""
                summary_selectors = ['p', '[class*="summary"]', '[class*="description"]', '[class*="excerpt"]']
                for selector in summary_selectors:
                    summary_elem = element.find(selector)
                    if summary_elem:
                        summary = clean_text(summary_elem.get_text(strip=True))
                        if summary:
                            break
                
                articles.append({
                    'category': category_name,
                    'category_display': category_display,
                    'title': title,
                    'link': link,
                    'date': date,
                    'summary': summary,
                    'full_content': '',
                    'image_url': ''
                })
                
            except Exception as e:
                continue
        
        return articles
    
    def fetch_full_content(article_url):
        soup = fetch_page(article_url)
        if not soup:
            return "", ""
        
        # Extract full content
        content = ""
        content_selectors = [
            'div[class*="story-body"]',
            'div[class*="article-body"]',
            'div[class*="content"]',
            'div[class*="story"]',
            'article',
            'main'
        ]
        
        for selector in content_selectors:
            content_elem = soup.select(selector)
            if content_elem:
                paragraphs = content_elem[0].find_all('p')
                if paragraphs:
                    content = ' '.join([clean_text(p.get_text(strip=True)) for p in paragraphs])
                    if content:
                        break
        
        if not content:
            article_main = soup.find('article') or soup.find('main')
            if article_main:
                paragraphs = article_main.find_all('p')
                if paragraphs:
                    content = ' '.join([clean_text(p.get_text(strip=True)) for p in paragraphs])
        
        # Extract image
        img_url = ""
        img_elem = soup.find('img')
        if img_elem:
            img_url = img_elem.get('src', '')
            if img_url and not img_url.startswith('http'):
                if img_url.startswith('//'):
                    img_url = 'https:' + img_url
                else:
                    img_url = 'https://www.bbc.com' + img_url
        
        return content, img_url
    
    # Scrape each category
    print("Starting BBC News scraping...")
    total_articles = 0
    
    for category in categories:
        print(f"Scraping: {category['display']}")
        soup = fetch_page(category['url'])
        if not soup:
            continue
        
        articles = extract_articles_from_page(soup, category['name'], category['display'])
        print(f"  Found {len(articles)} articles")
        
        # Fetch full content for each article
        for i, article in enumerate(articles, 1):
            print(f"  [{i}/{len(articles)}] Fetching content...")
            content, img = fetch_full_content(article['link'])
            article['full_content'] = content
            article['image_url'] = img
            time.sleep(0.3)
            all_articles.append(article)
        
        total_articles += len(articles)
        print(f"  Completed: {category['display']}")
        time.sleep(1)
    
    print(f"Total articles scraped: {total_articles}")
    return pd.DataFrame(all_articles)

# ==================== LOAD OR SCRAPE DATA ====================
def load_data():
    filename = "bbc_news_complete.csv"
    
    # Check if file exists, if not scrape
    if os.path.exists(filename):
        print(f"Loading data from {filename}")
        df = pd.read_csv(filename, encoding='utf-8-sig')
        
        # Parse date
        if 'date' in df.columns:
            try:
                df['date_parsed'] = pd.to_datetime(df['date'], errors='coerce')
            except:
                df['date_parsed'] = datetime.now()
        else:
            df['date_parsed'] = datetime.now()
    else:
        print("No existing data found. Scraping BBC News...")
        df = scrape_bbc_news()
        
        # Save for future use
        df.to_csv(filename, index=False, encoding='utf-8-sig')
        print(f"Data saved to {filename}")
    
    # Clean content
    patterns = [
        r'Photo credit.*?$',
        r'Credit:.*?$',
        r'Image credit.*?$',
        r'Getty Images.*?$',
        r'AP Photo.*?$',
        r'Reuters.*?$',
        r'AFP.*?$',
        r'Copyright.*?$',
        r'©.*?$',
        r'Share this.*?$',
        r'[0-9]+ comments.*?$',
    ]
    
    for col in ['full_content', 'summary', 'title']:
        if col in df.columns:
            for pattern in patterns:
                df[col] = df[col].str.replace(pattern, '', regex=True, flags=re.IGNORECASE)
            df[col] = df[col].str.replace(r'\s+', ' ', regex=True).str.strip()
            df[col] = df[col].str.replace(r'\n+', ' ', regex=True).str.strip()
    
    # Word count
    if 'full_content' in df.columns:
        df['word_count'] = df['full_content'].str.split().str.len().fillna(0).astype(int)
    else:
        df['word_count'] = 0
    
    # Remove duplicates
    if 'link' in df.columns:
        df = df.drop_duplicates(subset=['link'], keep='first')
    
    return df

# ==================== DASHBOARD ====================
def format_date_english(date_obj):
    """Format date in a readable English format"""
    if pd.isna(date_obj):
        return ""
    
    if isinstance(date_obj, str):
        try:
            date_obj = pd.to_datetime(date_obj)
        except:
            return date_obj
    
    try:
        return date_obj.strftime('%b %d, %Y')
    except:
        return str(date_obj)

# Load data
print("Loading data...")
df = load_data()

if df is None or df.empty:
    print("No data available!")
    exit()

# Clean date
df['date_parsed'] = pd.to_datetime(df['date_parsed'], errors='coerce')
df = df.dropna(subset=['date_parsed'])

if df.empty:
    print("No valid date data!")
    exit()

# Get min and max dates
min_date = df['date_parsed'].min()
max_date = df['date_parsed'].max()

# App
app = dash.Dash(__name__)

app.layout = html.Div([

    html.H2("📰 BBC News Dashboard", style={'textAlign': 'center', 'color': '#0d47a1'}),

    html.Div([
        # Date Range Filter
        html.Div([
            html.Label("📅 Date", style={'fontWeight': 'bold'}),
            dcc.DatePickerRange(
                id="date-range",
                start_date=min_date,
                end_date=max_date,
                display_format='YYYY-MM-DD',
                style={'width': '100%'}
            )
        ], style={'width': '45%', 'display': 'inline-block', 'padding': '10px', 'verticalAlign': 'top'}),

        # Category Filter
        html.Div([
            html.Label("📂 Category", style={'fontWeight': 'bold'}),
            dcc.Dropdown(
                id="category-filter",
                options=[{"label": c, "value": c} for c in df["category_display"].unique()] if 'category_display' in df.columns else [],
                multi=True,
                placeholder="Select Category",
                style={'width': '100%'}
            )
        ], style={'width': '45%', 'display': 'inline-block', 'padding': '10px', 'verticalAlign': 'top'})
    ], style={'padding': '20px', 'backgroundColor': '#f8f9fa', 'borderRadius': '10px'}),

    # Stats Row
    html.Div([
        html.Div(id="total-articles", style={'display': 'inline-block', 'padding': '20px', 'fontSize': '20px'}),
        html.Div(id="total-categories", style={'display': 'inline-block', 'padding': '20px', 'fontSize': '20px'}),
        html.Div(id="total-words", style={'display': 'inline-block', 'padding': '20px', 'fontSize': '20px'}),
    ], style={'textAlign': 'center', 'padding': '10px'}),

    # Charts
    html.Div([
        dcc.Graph(id="category-chart", style={'width': '48%', 'display': 'inline-block'}),
        dcc.Graph(id="trend-chart", style={'width': '48%', 'display': 'inline-block'})
    ]),

    # Articles Display
    html.Div([
        html.H3("📰 News Articles"),
        html.Div(id="articles-list")
    ], style={'padding': '20px'})
])

@app.callback(
    [
        Output("category-chart", "figure"),
        Output("trend-chart", "figure"),
        Output("articles-list", "children"),
        Output("total-articles", "children"),
        Output("total-categories", "children"),
        Output("total-words", "children")
    ],
    [
        Input("date-range", "start_date"),
        Input("date-range", "end_date"),
        Input("category-filter", "value")
    ]
)
def update_dashboard(start_date, end_date, categories):
    
    dff = df.copy()
    
    # Ensure datetime format
    dff["date_parsed"] = pd.to_datetime(dff["date_parsed"], errors="coerce")
    dff = dff.dropna(subset=['date_parsed'])
    
    # start_date filter
    if start_date is not None and start_date != "":
        start_date = pd.to_datetime(start_date)
        dff = dff[dff["date_parsed"] >= start_date]
    
    # end_date filter
    if end_date is not None and end_date != "":
        end_date = pd.to_datetime(end_date)
        dff = dff[dff["date_parsed"] <= end_date]
    
    # Category filter
    if categories and len(categories) > 0:
        dff = dff[dff["category_display"].isin(categories)]
    
    # If no data after filtering
    if dff.empty:
        fig_pie = px.pie(title="Articles by Category (No Data)")
        fig_line = px.line(title="Articles Over Time (No Data)")
        
        return (
            fig_pie, 
            fig_line, 
            [html.P("No articles found", style={'textAlign': 'center', 'padding': '20px'})],
            "📰 Articles: 0",
            "📂 Categories: 0",
            "📝 Words: 0"
        )
    
    # Stats
    total_articles = f"📰 Articles: {len(dff)}"
    total_categories = f"📂 Categories: {dff['category_display'].nunique()}"
    total_words = f"📝 Words: {dff['word_count'].sum():,}" if 'word_count' in dff.columns else "📝 Words: 0"
    
    # Category chart
    if 'category_display' in dff.columns and not dff.empty:
        cat_counts = dff['category_display'].value_counts()
        fig_pie = px.pie(
            values=cat_counts.values,
            names=cat_counts.index,
            title="Articles by Category"
        )
    else:
        fig_pie = px.pie(title="Articles by Category (No Data)")
    
    # Trend chart
    if 'date_parsed' in dff.columns and not dff.empty:
        daily_counts = dff.groupby(dff['date_parsed'].dt.date).size()
        fig_line = px.line(
            x=daily_counts.index,
            y=daily_counts.values,
            title="Articles Over Time",
            labels={'x': 'Date', 'y': 'Count'}
        )
    else:
        fig_line = px.line(title="Articles Over Time (No Data)")
    
    # Articles list
    articles_html = []
    
    # Sort by date (newest first)
    dff = dff.sort_values('date_parsed', ascending=False)
    
    # Colors for categories
    colors = {
        'World': '#FF6B6B', 'US & Canada': '#4ECDC4', 'UK': '#45B7D1',
        'Business': '#96CEB4', 'Technology': '#FFEAA7', 'Science': '#DDA0DD',
        'Health': '#98D8C8', 'Sport': '#F7DC6F', 'Entertainment': '#BB8FCE'
    }
    
    for _, row in dff.iterrows():
        # Get content
        content = row.get('full_content', row.get('summary', ''))
        if isinstance(content, str):
            content = content.strip()
        else:
            content = str(content)
        
        # Truncate
        if len(content) > 500:
            content = content[:500] + "..."
        
        category = row.get('category_display', '')
        color = colors.get(category, '#95A5A6')
        
        # Format date
        date_formatted = format_date_english(row.get('date_parsed'))
        
        # Get title
        title = row.get('title', 'No Title')
        if pd.isna(title):
            title = 'No Title'
        
        article_html = html.Div([
            html.H4(title, style={'color': '#0d47a1', 'marginBottom': '5px'}),
            html.Div([
                html.Span(category, style={
                    'backgroundColor': color, 
                    'color': 'white', 
                    'padding': '2px 10px', 
                    'borderRadius': '20px',
                    'fontSize': '12px',
                    'marginRight': '10px'
                }),
                html.Span(f"📅 {date_formatted}", style={'marginRight': '10px'}),
                html.Span(f"📊 {row.get('word_count', 0)} words", style={'marginRight': '10px'}),
                html.A("🔗 Link", href=row.get('link', '#'), target='_blank', style={'color': '#0d47a1'})
            ], style={'marginBottom': '10px'}),
            html.P(content if content else 'No content available', style={'color': '#333', 'lineHeight': '1.6'}),
            html.Hr()
        ], style={
            'backgroundColor': 'white', 
            'padding': '15px', 
            'borderRadius': '10px',
            'boxShadow': '0 2px 8px rgba(0,0,0,0.1)',
            'marginBottom': '15px',
            'borderLeft': f'5px solid {color}'
        })
        
        articles_html.append(article_html)
    
    return fig_pie, fig_line, articles_html, total_articles, total_categories, total_words

if __name__ == "__main__":
    app.run(debug=True)
```
