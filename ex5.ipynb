```python
import requests
from bs4 import BeautifulSoup
import csv

html_doc = requests.get('https://www.worldometers.info/coronavirus/')
soup = BeautifulSoup(html_doc.content, 'html.parser')

table_content = soup.find('table')

table_head = table_content.find('thead')
table_body = table_content.find('tbody')
table_row = table_body.find_all('tr')
headerrow=table_head.find_all('th')
headers_list = [th.text.strip() for th in headerrow]

file=open("yoniy12345679.csv",'w',newline='',encoding='utf-8')
write= csv.writer(file)
write.writerow(headers_list)
# data_rows = []
for row in table_row[8:]:
    cols = row.find_all('td')
    
    # Clean each cell: strip whitespace, remove newlines but keep commas inside numbers
    # cleaned_cells = [col.text.strip() for col in cols]
    # data_rows.append(cleaned_cells)
    data_rows = []
    for col in cols:
        data_rows.append(col.text.strip().replace(",",""))
       
       
    write.writerow(data_rows)
        
    # write.writerow(cleaned_cells)

file.close()    
print("hello")
```
