```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


df = pd.read_excel("COVID-19-geographic-disbtribution-worldwide.xlsx")


africa = df[df["continentExp"] == "Africa"]


plt.figure(figsize=(8,6))
sns.scatterplot(
    data=africa,
    x="cases",
    y="deaths",
    hue="countriesAndTerritories"
)

plt.title("COVID-19 in Africa")
plt.xlabel("Cases")
plt.ylabel("Deaths")
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_excel("COVID-19-geographic-disbtribution-worldwide.xlsx")
africa = df[df["continentExp"] == "Africa"]

highest_5 = africa.nlargest(5, "cases")
smallest_5 = africa.nsmallest(5, "cases")
data = pd.concat([highest_5, smallest_5])

sns.scatterplot(
    data=data,
    x="cases",
    y="deaths",
    hue="countriesAndTerritories",
    s=120
)

plt.title("Top 5 Highest and Lowest COVID-19 Records in Africa")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_excel("COVID-19-geographic-disbtribution-worldwide.xlsx")

# Group by continent
continent_data = df.groupby("continentExp")[["cases", "deaths"]].sum().reset_index()

print(continent_data)

# Scatter plot
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=continent_data,
    x="cases",
    y="deaths",
    hue="continentExp",
    s=200
)

plt.title("COVID-19 Cases vs Deaths by Continent")
plt.xlabel("Total Cases")
plt.ylabel("Total Deaths")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_excel("COVID-19-geographic-disbtribution-worldwide.xlsx")

# Group by continent
continent_data = df.groupby("continentExp")[["cases", "deaths"]].sum().reset_index()

# Plot
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=continent_data,
    x="cases",
    y="deaths",
    hue="continentExp",
    s=200
)

# Show continent names on points
for i in range(len(continent_data)):
    plt.text(
        continent_data["cases"][i],
        continent_data["deaths"][i],
        continent_data["continentExp"][i]
    )

plt.title("COVID-19 Cases and Deaths by Continent")
plt.xlabel("Total Cases")
plt.ylabel("Total Deaths")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_excel("COVID-19-geographic-disbtribution-worldwide.xlsx")

# Replace popData2019 with your actual population column name
continent_data = df.groupby("continentExp").agg({
    "cases": "sum",
    "deaths": "sum",
    "popData2019": "sum"
}).reset_index()

print(continent_data)

# Scatter plot
sns.scatterplot(
    data=continent_data,
    x="popData2019",
    y="deaths",
    hue="continentExp",
    s=200
)

plt.title("Population vs Deaths by Continent")
plt.xlabel("Population")
plt.ylabel("Deaths")
plt.show()
```

```python

from dash import Dash, dcc, html, Input, Output
import plotly.express as px
app = Dash(__name__)
app.layout = html.Div([ html.H1('Restaurant tips by day of week'),
                        dcc.Dropdown( id="dropdown", options=['Mon', "Fri", "Sat", "Sun"],
                                      value="Fri",
                                      clearable=False,),
                             dcc.Graph(id="graph"),
])
@app.callback(
 Output("graph", "figure"),
 Input("dropdown", "value"))
def update_bar_chart(day):
 df = px.data.tips() # replace with your own data source
 mask = df["day"] == day
 fig = px.bar(df[mask], x="sex", y="total_bill", color="smoker", barmode="group")
 return fig
app.run(debug=True)
```
