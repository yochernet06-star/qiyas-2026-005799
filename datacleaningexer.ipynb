```python
import numpy as np
import pandas as pd
data=({
    "name":["abel","sara",None,"john"],
    "age":[20,np.nan,25,30],
    "salary":[4000,4500,5000,45000],
    "join_date":[
        "2026-01-01",
        "01/02/2026",
        "March 10,2026",
        "2026.04.15"]
})
df=pd.DataFrame(data)
```

```python
df
```

```python
df_cleaned = df.dropna()
print(df_cleaned)
```

```python
for column in df.columns:
 if df[column].dtype == 'float64' or df[column].dtype == 'int64':
  df[column].fillna(df[column].mean())
df["name"]=df["name"].fillna("unkown")
print(df)
```

```python
from dateutil.parser import parse
def parse_date(date_str):
 try:
  return parse(date_str, dayfirst=False)
 except ValueError:
  return pd.NaT
df['join_date'] = df['join_date'].apply(parse_date)
df['join_date'] = df['join_date'].dt.strftime('%Y-%m-%d')
print("Result")
print(df)
```

```python
df = pd.DataFrame(data)
Q1 = df['salary'].quantile(0.25)
Q3 = df['salary'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
df['Outlier Column'] = df['salary'].apply(lambda x: 'Outlier' if x > upper_bound else 'Normal')
```
